# Engenharia de variáveis

## Defasagem e Médias

### Ordenação Cronológica dos Dados

```python
df_resultados_com_clima = df_resultados_com_clima.sort_values(by=['driver_key', 'ano', 'rodada']).reset_index(drop=True)
#display(df_resultados_com_clima.head(5))

df_qualy_features       = df_qualy_features.sort_values(by=['driver_key', 'ano', 'rodada']).reset_index(drop=True)
#display(df_qualy_features.head(5))
```

### Criação das Variáveis base

```python
df_resultados_com_clima['total_voltas_corrida']     = (df_resultados_com_clima.groupby('race_key')['voltas_completadas'].transform('max'))
df_resultados_com_clima['taxa_voltas_completadas']  = (df_resultados_com_clima['voltas_completadas']/df_resultados_com_clima['total_voltas_corrida'])
#display(df_resultados_com_clima.head(5))
```

### Criação das Colunas de Defasagem e Médias

#### df_resultados_com_clima

```python
df_resultados_com_clima['taxa_voltas_completadas_media_7']  = (df_resultados_com_clima
                                                               .groupby('driver_key')['taxa_voltas_completadas']
                                                               .shift(1)
                                                               .rolling(window=7, min_periods=1)
                                                               .mean())

for n in range(1,8):
  df_resultados_com_clima[f'posicao_final_lag_{n}']         = df_resultados_com_clima.groupby('driver_key')['posicao_final'].shift(n)

df_resultados_com_clima['posicao_final_media_7']            = (df_resultados_com_clima
                                                               .groupby('driver_key')['posicao_final']
                                                               .shift(1)
                                                               .rolling(window=7, min_periods=1)
                                                               .mean())

df_resultados_com_clima['posicao_final_media_7_lag_1']      = df_resultados_com_clima.groupby('driver_key')['posicao_final_media_7'].shift(1)

df_resultados_com_clima['delta_tendencia_posicao_final']    = df_resultados_com_clima['posicao_final_media_7'] - df_resultados_com_clima['posicao_final_media_7_lag_1']

df_resultados_com_clima['is_overtake_media_7']              = (df_resultados_com_clima
                                                               .groupby('driver_key')['is_overtake']
                                                               .shift(1)
                                                               .rolling(window=7, min_periods=1)
                                                               .mean())

#display(df_resultados_com_clima.head(5))
```

#### df_qualy_features

```python
for n in range (1,8):
  df_qualy_features[f'posicao_grid_lag_{n}']    = df_qualy_features.groupby('driver_key')['posicao_grid'].shift(n)

df_qualy_features['posicao_grid_media_7']       = (df_qualy_features
                                                   .groupby('driver_key')['posicao_grid']
                                                   .shift(1)
                                                   .rolling(window=7, min_periods=1)
                                                   .mean())

df_qualy_features['posicao_grid_media_7_lag_1'] = df_qualy_features.groupby('driver_key')['posicao_grid_media_7'].shift(1)

df_qualy_features['delta_tendencia_grid']       = df_qualy_features['posicao_grid_media_7'] - df_qualy_features['posicao_grid_media_7_lag_1']

tempo_pole_gp = df_qualy_features.groupby('race_key')['q_avg_seconds'].transform('min')

df_qualy_features['q_gap_seconds']              = df_qualy_features['q_avg_seconds'] - tempo_pole_gp

pior_gap_gp = df_qualy_features.groupby('race_key')['q_gap_seconds'].transform('max')

df_qualy_features['q_gap_media_7']              = (df_qualy_features
                                                   .groupby('driver_key')['q_gap_seconds']
                                                   .shift(1)
                                                   .rolling(window=7, min_periods=1)
                                                   .mean())


df_qualy_features['q_gap_media_7_lag_1'] = df_qualy_features.groupby('driver_key')['q_gap_media_7'].shift(1)
df_qualy_features['delta_tendencia_q_gap'] = df_qualy_features['q_gap_media_7'] - df_qualy_features['q_gap_media_7_lag_1']

#display(df_qualy_features.head(5))
```

## Merge

```python
## Merge

colunas_para_trazer = [
    "race_key",
    "driver_key",
    "posicao_final",
    "taxa_voltas_completadas",
    "taxa_voltas_completadas_media_7",
    "posicao_final_media_7",
    "posicao_final_media_7_lag_1",
    "delta_tendencia_posicao_final",
    "is_overtake_media_7"
] + [f"posicao_final_lag_{n}" for n in range(1, 8)]

df_f1_modelo = pd.merge(
    df_qualy_features,
    df_resultados_com_clima[colunas_para_trazer],
    on=["race_key", "driver_key"],
    how="inner"
)

df_f1_modelo["is_top_10"] = (df_f1_modelo["posicao_final"] <= 10).astype(int)

df_f1_treino_pronto = df_f1_modelo.dropna(
    subset=["posicao_final_lag_7", "posicao_grid_lag_7", "q_gap_seconds"]
).reset_index(drop=True)

display(df_f1_treino_pronto.sample(5))
```

## Preparação do treino

### Definição Formal das Variáveis

```python
# 1. Variável Alvo / Gabarito (y)
target = "is_top_10"

# 2. Variáveis Categóricas Estruturais (One-Hot Encoding)
cols_categoricas = ["constructor_key", "circuit_key"]

# 3. Variáveis Numéricas Preditivas (X)
cols_numericas = [
    # Condição pré-corrida do GP atual (Sábado e Clima)
    "posicao_grid",
    "q_gap_seconds",
    "idade_piloto_corrida",
    "wet", "dry", "warm", "cold", "windy", "cloudy",

    # Retrospecto de Qualificação / Sábado (Janela de 7 etapas)
    "posicao_grid_media_7",
    "posicao_grid_media_7_lag_1",
    "delta_tendencia_grid",
    "q_gap_media_7",
    "q_gap_media_7_lag_1",
    "delta_tendencia_q_gap",

    # Retrospecto de Corrida / Domingo (Janela de 7 etapas)
    "posicao_final_media_7",
    "posicao_final_media_7_lag_1",
    "delta_tendencia_posicao_final",
    "taxa_voltas_completadas_media_7",
    "is_overtake_media_7"
] + [f"posicao_grid_lag_{n}" for n in range(1, 8)] + [f"posicao_final_lag_{n}" for n in range(1, 8)]
```

### Codificação de Categóricas (One-Hot Encoding)

```python
# One-Hot Encoding nas colunas de equipe e circuito
df_encoded = pd.get_dummies(
    df_f1_treino_pronto,
    columns=cols_categoricas,
    dtype=int
)

# Identificamos automaticamente o nome de todas as novas colunas binárias criadas
novas_cols_categoricas = [col for col in df_encoded.columns if col.startswith("constructor_key_") or col.startswith("circuit_key_")]

# Lista completa e definitiva de features (X)
todas_features = cols_numericas + novas_cols_categoricas
```

### Divisão Cronológica dos Dados

```python
# 1. Definição das máscaras temporais
mask_treino    = df_encoded["ano"] <= 2024
mask_teste     = df_encoded["ano"] == 2025
mask_valid2026 = df_encoded["ano"] == 2026

# 2. Separação em X (atributos de entrada) e y (gabarito)
X_train_raw = df_encoded.loc[mask_treino, todas_features].copy()
y_train     = df_encoded.loc[mask_treino, target].values

X_test_raw  = df_encoded.loc[mask_teste, todas_features].copy()
y_test      = df_encoded.loc[mask_teste, target].values

X_2026_raw  = df_encoded.loc[mask_valid2026, todas_features].copy()
y_2026      = df_encoded.loc[mask_valid2026, target].values
```

### Escalonamento Numérico (MinMaxScaler)

```python
from sklearn.preprocessing import MinMaxScaler

# 1. Instanciamos o normalizador para a escala de 0 a 1
scaler = MinMaxScaler(feature_range=(0, 1))

# 2. Ajustamos (fit) a escala no treino e já transformamos os dados de treino
X_train_num_scaled = scaler.fit_transform(X_train_raw[cols_numericas])

# 3. Aplicamos a mesma régua (transform) no teste e na validação de 2026 (SEM dar fit)
X_test_num_scaled  = scaler.transform(X_test_raw[cols_numericas])
X_2026_num_scaled  = scaler.transform(X_2026_raw[cols_numericas])

# 4. Reconstruímos os DataFrames finais combinando as numéricas escaladas com as binárias
import numpy as np

X_train = np.hstack([X_train_num_scaled, X_train_raw[novas_cols_categoricas].values])
X_test  = np.hstack([X_test_num_scaled,  X_test_raw[novas_cols_categoricas].values])
X_2026  = np.hstack([X_2026_num_scaled,  X_2026_raw[novas_cols_categoricas].values])

# 5. Conferência de integridade dos arrays
print(f"X_train shape: {X_train.shape} | y_train shape: {y_train.shape}")
print(f"X_test shape:  {X_test.shape}  | y_test shape:  {y_test.shape}")
print(f"X_2026 shape:  {X_2026.shape}  | y_2026 shape:  {y_2026.shape}")
```
